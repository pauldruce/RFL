# RFL Playground

Welcome to the Random Fuzzy Library playground! This notebook demonstrates how to interact with the C++ core using the Python bindings.

```python
import rfl
import numpy as np
import matplotlib.pyplot as plt

# Initialise a Dirac Operator
p, q = 1, 3
dim = 10
dirac = rfl.DiracOperator(p, q, dim)

print(f"Dirac Operator Type: {dirac.get_type()}")
print(f"Matrix Dimension: {dirac.get_matrix_dimension()}")
```

```python
# Setup the Metropolis Algorithm
g_2 = -1.0
g_4 = 1.0
scale = 1.0
num_steps = 100
seed = 42

metropolis = rfl.Metropolis(g_2, g_4, scale, num_steps, seed)
acceptance_rate = metropolis.update_dirac(dirac)

print(f"Acceptance Rate: {acceptance_rate * 100:.2f}%")
```

```python
# Analyse Eigenvalues
eigenvals = dirac.get_eigenvalues()

plt.figure(figsize=(8, 6))
plt.hist(eigenvals, bins=30, alpha=0.7, color='blue', edgecolor='black')
plt.title('Dirac Operator Eigenvalue Distribution')
plt.xlabel('Eigenvalue')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```
